```python
import numpy as np

# 1 & 2: filter Asia cars, select the two columns
lr = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']]

# 3: first 7 rows
lr = lr.iloc[:7]

# 4: get the NumPy array
X = lr.to_numpy()

# 5: X transpose times X
XTX = X.T @ X

# 6: invert XTX
XTX_inv = np.linalg.inv(XTX)

# 7: create y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8: w = XTX_inv @ X.T @ y
w = XTX_inv @ X.T @ y

# 9: sum of elements of w
print(w.sum())
```
